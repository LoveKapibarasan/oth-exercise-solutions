## A. Propagate a probability law

On a probability space $(\Omega,\mathcal F,\Pr)$, let $X_t:\Omega\to D$
be random variables with finite state space $D=\{0,1\}$, whose subsets
are all measurable, and times
$t\in\mathbb N_0=\{0,1,2,\ldots\}$. A time-homogeneous Markov chain has
a fixed transition matrix $P=(P_{ij})_{i,j\in D}$ such that, for every
time $t\in\mathbb N_0$, history $(i_0,\ldots,i_t)\in D^{t+1}$ with
positive probability, and next state $j\in D$,
$$\Pr(X_{t+1}=j\mid X_0=i_0,\ldots,X_t=i_t)=P_{i_tj}.$$
Here $P_{ij}\geq0$ and $\sum_{j\in D}P_{ij}=1$: each row distributes
probability over all next states for one fixed current state. Columns use
different conditioning states, so their unweighted sums need not equal 1.
A positive-probability history means
$\Pr(X_0=i_0,\ldots,X_t=i_t)>0$, making its conditional-probability ratio
defined. The **Markov property** says that at a specified time the current
state suffices for the conditional distribution of the next state;
**time homogeneity** additionally says the same matrix applies at every time.
Neither property asserts independence of consecutive states. The worked
cooling reference below separates these two properties.

The **row vector** $\boldsymbol\mu_t=(\mu_t(0),\mu_t(1))$, where
$\mu_t(j)=\Pr(X_t=j)$ for $j\in D$, is the
law at a fixed time, not a pair of observed visit fractions. For $t\geq0$,
$$\boldsymbol\mu_{t+1}=\boldsymbol\mu_tP,\qquad
  \boldsymbol\mu_t=\boldsymbol\mu_0P^t,\qquad P^0=I_2,$$
where $I_2$ is the two-by-two identity matrix. In this demonstration,
$$P=\begin{pmatrix}0.8&0.2\\0.3&0.7\end{pmatrix},\qquad
  \boldsymbol\pi=(0.6,0.4).$$

**Why multiply a law by a matrix?** For events $H,B$ with $\Pr(H)>0$,
$\Pr(H\cap B)=\Pr(H)\Pr(B\mid H)$. The cases $\{X_t=0\}$ and
$\{X_t=1\}$ partition the possible current states. Total probability gives
$$\mu_{t+1}(j)=\sum_{i\in D}\mu_t(i)P_{ij}\qquad(j\in D).$$
We multiply within each current-to-next-state case and add across cases;
independence is not assumed. A state with $\mu_t(i)=0$ contributes zero,
without assigning an elementary conditional ratio to that null event.
This is exactly the component formula for $\boldsymbol\mu_tP$.

**Connection with Session 5.** The proposal law $q(y\mid x)$ describes the
feasible tentative state **before acceptance**; $\alpha_T(x,y)$ is its
conditional acceptance probability. At fixed temperature $T>0$, $K_T(x,y)$
is the probability of the actual next state after both stages, including
all ways of staying. A fixed objective, proposal rule, and temperature, with
fresh independent inputs, produce a homogeneous Markov chain: its transition
matrix has entries $P_{ij}=K_T(i,j)$, **not** $q(j\mid i)$.

In Session 5's five-state example, from state 4 at $T=2$,
$$q(3\mid4)=\tfrac12,\qquad \alpha_2(4,3)=e^{-1},\qquad
  K_2(4,3)=\tfrac12 e^{-1}\approx0.183940.$$
Symmetry of $q$ does not imply symmetry of the complete matrix $K_T$ or $P$.
The two-state $P$ displayed above is a **new example specified directly**.
It is not a relabelled or rounded version of Session 5's five-state $K_T$.
Its entries already describe complete transitions: do not multiply them
by another acceptance probability.

A probability row vector has nonnegative entries summing to one.
Such a vector $\boldsymbol\pi$ is **invariant** when
$\boldsymbol\pi P=\boldsymbol\pi$. **Detailed balance** means
$\pi_iP_{ij}=\pi_jP_{ji}$ for every $i,j\in D$; summing over $i$ proves
invariance. Here the two directed off-diagonal flows both equal $0.12$.
These checks use the **complete** transition matrix $P$ (or $K_T$ in a Metropolis
construction), not the proposal law $q$ alone.

**Supplied reading connection: Session 5's five-state update.**
This reference supports S6.2, counted once; it adds no exercise or
simulation. The full proof is in the Session 6 notes. Let $D_5=\{0,1,2,3,4\}$ and $f:D_5\to\mathbb R$ have
values $(0,3,1,4,2)$ in state order. Use Session 5's symmetric proposal law
$q$ and complete transition $K_T$, at fixed $T>0$. Define
$$Z_T=\sum_{z\in D_5}e^{-f(z)/T}\in(0,\infty),\qquad
p_T(x)=\frac{e^{-f(x)/T}}{Z_T}\quad(x\in D_5).$$
These positive masses define the law
$\pi_T(A)=\sum_{x\in A}p_T(x)$ for $A\subseteq D_5$ and its probability
row vector $\boldsymbol p_T=(p_T(x))_{x\in D_5}$.

**Optional algebra: the general symmetric-proposal identity.**
For distinct $x,y$, symmetry gives
$$p_T(x)K_T(x,y)=q(y\mid x)\min\{p_T(x),p_T(y)\}
                =p_T(y)K_T(y,x).$$
For $x=y$, the identity is automatic. Thus detailed balance holds for all
pairs, including the diagonal with its rejection mass.

**Supplied five-state example supporting S6.2.** At $T=2$:

| Direction | Current mass | Actual transition | Probability flow |
| --- | --- | --- | --- |
| $4\to3$ | $e^{-1}/Z_2$ | $\tfrac12e^{-1}$ | $e^{-2}/(2Z_2)$ |
| $3\to4$ | $e^{-2}/Z_2$ | $\tfrac12$ | $e^{-2}/(2Z_2)$ |

These flows are probabilities of opposite one-step events **when the current
law is $\pi_2$**. Equal flows do not require equal transition probabilities.
Checking this pair alone would not establish the all-pairs identity.

Using detailed balance for all pairs, let $X$ have law $\pi_T$ and let
$\Pr(X^+=y\mid X=x)=K_T(x,y)$ for all $x,y\in D_5$. For fixed $y\in D_5$,
$$\Pr(X^+=y)=\sum_{x\in D_5}p_T(x)K_T(x,y)
 =p_T(y)\sum_{x\in D_5}K_T(y,x)=p_T(y).$$
The steps use total probability, all-pairs detailed balance and factoring,
then the sum of **row $y$**, whose next-state index $x$ ranges over $D_5$.
Equivalently the probability row vector satisfies
$\boldsymbol p_TK_T=\boldsymbol p_T$: the law $\pi_T$ is invariant.
This preserves a law already present; it does not prove convergence from
another initial law.

The supplied code weights each row of `K_five` by `p_five` to form
`flows_five`, then separately calculates `p_five @ K_five`. The first
array contains joint transition probabilities under $\pi_2$; the second
is the next-state probability row vector. The printout's `p2` denotes the
mass vector $\boldsymbol p_2$ representing law $\pi_2$. Numerical checks allow roundoff and do not replace
the proof. This reference uses the known objective and no random inputs.

Return to the **two-state** $P$ for the guided calculation.
Predict $\boldsymbol\mu_1$ and $\boldsymbol\mu_2$ from fixed start $X_0=0$.
Explain why $P_{01}=0.2$ is already an actual transition probability,

The function `exact_laws` propagates the law by matrix multiplication,
without random sampling; its floating-point results can have roundoff.
With `law_horizon=16`, each returned array has shape `(17, 2)`: row `t`
is the law at time $t=0,\ldots,16$, and columns 0 and 1 are the two state
probabilities. Thus `laws_start0[t, 1]` represents $\mu_t(1)$ from start 0.
Each graph marker is a probability at an integer time, not an observed
state. Connecting segments only guide the eye; the model has no states
at intermediate noninteger times. The dashed line is the invariant mass.

```python
import numpy as np
import matplotlib.pyplot as plt

P = np.array([[0.8, 0.2], [0.3, 0.7]])
pi = np.array([0.6, 0.4])

def validate_matrix(matrix):
    matrix = np.asarray(matrix, dtype=float)
    if (matrix.shape != (2, 2) or not np.all(np.isfinite(matrix))
            or np.any(matrix < 0) or np.any(matrix > 1)
            or not np.allclose(matrix.sum(axis=1), 1, rtol=0, atol=1e-12)):
        raise ValueError('Use a finite nonnegative 2-by-2 matrix with row sums 1.')
    return matrix

def exact_laws(matrix, initial_law, steps):
    matrix = validate_matrix(matrix)
    initial_law = np.asarray(initial_law, dtype=float)
    if (initial_law.shape != (2,) or not np.all(np.isfinite(initial_law))
            or np.any(initial_law < 0) or np.any(initial_law > 1)
            or not np.isclose(initial_law.sum(), 1, rtol=0, atol=1e-12)):
        raise ValueError('Use a probability row vector of length 2.')
    if not isinstance(steps, (int, np.integer)) or steps < 0:
        raise ValueError('The number of transitions must be a nonnegative integer.')
    laws = np.empty((steps+1, 2))
    laws[0] = initial_law
    for t in range(steps):
        laws[t+1] = laws[t] @ matrix
    return laws

law_horizon = 16
laws_start0 = exact_laws(P, [1, 0], law_horizon)
laws_start1 = exact_laws(P, [0, 1], law_horizon)
for t in range(3):
    print(f't={t}, law from fixed start 0: {laws_start0[t]}')
print('Invariant law times P:', pi @ P)
flows = pi[:, None]*P
print('Stationary probability flows pi_i P_ij:\n', flows)
assert np.allclose(laws_start0[1:3], [[0.8, 0.2], [0.7, 0.3]])
assert np.allclose(pi @ P, pi)
assert np.allclose(flows, flows.T)

# Supplied five-state reference from Session 5; no simulation or extra exercise.
five_objective = np.array([0., 3., 1., 4., 2.])
Q_five = 0.5*np.array([[1, 1, 0, 0, 0], [1, 0, 1, 0, 0],
                      [0, 1, 0, 1, 0], [0, 0, 1, 0, 1], [0, 0, 0, 1, 1]])
alpha_five = np.exp(-np.maximum(five_objective[None, :]-five_objective[:, None], 0)/2.)
K_five = Q_five*alpha_five
np.fill_diagonal(K_five, 0.)
np.fill_diagonal(K_five, 1.-K_five.sum(axis=1))
p_five = np.exp(-five_objective/2.)
p_five /= p_five.sum()
flows_five = p_five[:, None]*K_five
p_five_after = p_five @ K_five
assert np.all(K_five >= 0) and np.allclose(K_five.sum(axis=1), 1.)
assert np.allclose(flows_five, flows_five.T)
assert np.allclose(p_five_after, p_five)
print('Supplied five-state reference at T=2, under current law p2')
print('x y     p2(x)   K2(x,y) p2(x)K2(x,y)')
for x_five, y_five in ((4, 3), (3, 4)):
    print(f'{x_five} {y_five} {p_five[x_five]:9.6f} '
          f'{K_five[x_five, y_five]:9.6f} {flows_five[x_five, y_five]:12.6f}')
with np.printoptions(precision=6, suppress=True):
    print('p2       =', p_five)
    print('p2 @ K2  =', p_five_after)

fig_laws, ax = plt.subplots(figsize=(7, 3.5))
times = np.arange(law_horizon+1)
ax.plot(times, laws_start0[:, 1], 'o-', label='Fixed start 0')
ax.plot(times, laws_start1[:, 1], 's-', label='Fixed start 1')
ax.axhline(pi[1], color='black', linestyle='--', label='Invariant mass of state 1')
ax.set(xlabel='Time t', ylabel='Exact probability Pr(X_t = 1)', ylim=(-0.02, 1.02))
ax.legend()
fig_laws.tight_layout()
plt.show()
```

## B. Simulate two paths; distinguish stationarity from convergence

For this matrix, write $a=P_{01}=0.2$, $b=P_{10}=0.3$, and
$r_t=\mu_t(1)=\Pr(X_t=1)$. The recurrence and its solution are
$$r_{t+1}=a+(1-a-b)r_t,\qquad
  r_t=\frac{a}{a+b}+\left(r_0-\frac{a}{a+b}\right)(1-a-b)^t.$$
Because $|1-a-b|=0.5<1$, the laws tend to
$\boldsymbol\pi=(b/(a+b),a/(a+b))=(0.6,0.4)$ from either fixed start.
This means $\mu_t(0)\to0.6$ and $\mu_t(1)\to0.4$ as $t\to\infty$.
The algebra proves convergence for this example; a plot alone does not.

A process is **stationary** when every finite joint law is unchanged by
a common nonnegative integer time shift. A homogeneous chain initialized with
$\boldsymbol\mu_0=\boldsymbol\pi$ is stationary. Our fixed-start chains
are not stationary: their initial laws differ from $\boldsymbol\pi$.
Even in stationarity, neighboring states are dependent here:
$\Pr(X_{t+1}=1\mid X_t=1)=0.7\ne0.4=\Pr(X_{t+1}=1)$.

For $n\in\{1,\ldots,N\}$, $N=2000$, the **random visit fraction** is
$$V_n:\Omega\to\{0,1/n,\ldots,1\},\qquad
V_n(\omega)=\frac1n\sum_{t=1}^n\mathbf1_{\{1\}}(X_t(\omega)).$$
Each indicator is 1 when $X_t(\omega)=1$ and 0 otherwise. It is a
measurable random variable; its finite average $V_n$ is therefore also
measurable. The notation $V_n=n^{-1}\sum_{t=1}^n\mathbf1_{\{X_t=1\}}$
suppresses the argument $\omega$, but still denotes a random variable.
For one realized path $x_t=X_t(\omega)$, its **observed value** is
$$v_n=V_n(\omega)=\frac1n\sum_{t=1}^n\mathbf1_{\{1\}}(x_t).$$
We store $x_0$ as array entry 0 and **exclude it** from the fraction, while
counting every post-transition repetition. The state array has $N+1$ entries.
$V_n$ averages times along one path; $r_t=\Pr(X_t=1)$ is a probability at
one fixed time. This $r_t$ is unrelated to the rejection mass $r_{T,s}(x)$
in the continuous example.

The model uses an infinite iid sequence $(U_t)_{t\geq1}$ with
$U_t\sim\mathrm{Uniform}(0,1)$, independent of $X_0$; the code generates
its first $N$ inputs (pseudorandom numbers in `[0, 1)`). Endpoint differences
have probability zero in the ideal continuous model.
Given $X_{t-1}=i$, set $X_t=0$ if $U_t<P_{i0}$, and $X_t=1$ otherwise.
This threshold samples the **actual next state** from row $i$ of $P$; it is
not Session 5's accept/reject test. For example, from state 1, $u=0.2<0.3$
gives next state 0. A stay is a valid next-state outcome, but this direct-P
example specifies no separate proposal or acceptance stage: its cause
cannot be identified as proposal self-loop or rejection.
Two independently generated sets of uniforms define independent paths
from the two fixed starts. `SeedSequence(6060).spawn(2)` supplies two
pseudorandom child streams to approximate this model; it does not prove
independence of deterministic computer output. Two homogeneous Markov-chain
constructions with the same $P$ and initial law give the same distribution
of state paths, but need not give the same realized path from the same seed:
they may consume or map random inputs differently.

The rule uses the **previous state**, so independent uniforms need not
produce independent output states. For example, start 0 with supplied
inputs `(0.9, 0.2, 0.85, 0.5)` gives states `(0, 1, 0, 1, 1)` and
$v_4=3/4$. By contrast the exact time-4 law gives $\mu_4(1)=0.375$.
These are an arithmetic check and a probability calculation, respectively.

The graph now plots $v_n$ along each single path. Its horizontal axis is
logarithmic: equal horizontal distances represent equal ratios of $n$.
The vertical axis includes all possible fractions from 0 to 1, with a
small margin. Consecutive points average overlapping prefixes and are
not independent repeated estimates. The final point summarizes all
$N=2000$ post-transition states; the dashed line marks the invariant mass,
not a guaranteed finite-path value.

Predict whether either finite path must give $v_N=0.4$ exactly. A finite
path gives an observed fraction, not an exact invariant mass. This graph
does not prove a limit theorem for time averages; later sessions develop
that theory. Do not apply Session 4's iid standard-error formula to these
dependent indicator observations.

```python
def simulate_two_state(matrix, start, uniforms):
    matrix = validate_matrix(matrix)
    uniforms = np.asarray(uniforms, dtype=float)
    if start not in (0, 1):
        raise ValueError('The start must be state 0 or state 1.')
    if (uniforms.ndim != 1 or not np.all(np.isfinite(uniforms))
            or np.any(uniforms < 0) or np.any(uniforms >= 1)):
        raise ValueError('Use a one-dimensional array of values in [0,1).')
    states = np.empty(len(uniforms)+1, dtype=int)
    states[0] = start
    for t, u in enumerate(uniforms, 1):
        states[t] = 0 if u < matrix[states[t-1], 0] else 1
    return states

def visit_fractions(states):
    # Entry 0 is the supplied initial state, not one of the N observations.
    observations = np.asarray(states)[1:]
    return np.cumsum(observations == 1)/np.arange(1, len(observations)+1)

transition_count = 2000
seed = 6060
children = np.random.SeedSequence(seed).spawn(2)
uniform_inputs = [np.random.default_rng(child).random(transition_count) for child in children]
paths = [simulate_two_state(P, start, uniform_inputs[start]) for start in (0, 1)]
fractions = [visit_fractions(path) for path in paths]
for start in (0, 1):
    print(f'Start {start}: first 13 states including x_0 = {paths[start][:13]}')
    print(f'          v_{transition_count} = {fractions[start][-1]:.4f}; invariant mass = {pi[1]:.4f}')
assert all(len(path) == transition_count+1 for path in paths)

fig_visits, ax = plt.subplots(figsize=(7, 3.5))
sample_counts = np.arange(1, transition_count+1)
for start in (0, 1):
    ax.plot(sample_counts, fractions[start], label=f'Observed path from start {start}')
ax.axhline(pi[1], color='black', linestyle='--', label='Invariant mass of state 1')
ax.set(xlabel='Observed transitions n (initial state excluded)', ylabel='Observed visit fraction v_n',
       xscale='log', ylim=(-0.02, 1.02))
ax.legend()
fig_visits.tight_layout()
plt.show()
```

## C. An invariant law does not ensure convergence of fixed-time laws

Let
$$F=\begin{pmatrix}0&1\\1&0\end{pmatrix},\qquad
  \boldsymbol\pi^F=(1/2,1/2),$$
where $\boldsymbol\pi^F$ is the invariant probability row vector for $F$.
We have $\boldsymbol\pi^F F=\boldsymbol\pi^F$, and detailed balance holds.
From fixed start 0, the path is $0,1,0,1,\ldots$, so the law is
$(1,0)$ at even times and $(0,1)$ at odd times. These laws do **not**
converge to $\boldsymbol\pi^F$. If initialized with law $\boldsymbol\pi^F$
instead, the process is stationary but still alternates at every transition.

For a finite state space $D$, a transition matrix $P$ is **irreducible** if
for every $i,j\in D$ there is an integer $k\geq0$ with $(P^k)_{ij}>0$.
Both $P$ and $F$ are irreducible. For state $i\in D$ in the flip example,
its **period** is the greatest common divisor of the nonempty set of positive
return times $\{k\in\{1,2,\ldots\}:(F^k)_{ii}>0\}$.
This set contains exactly the positive even integers, so both periods are 2.
A state is aperiodic when its period is 1.
The positive diagonal entries of $P$ give period 1 for the earlier example.
Reading the two rows of $F$ gives $X_{t+1}=1-X_t$ almost surely: from 0,
choose 1 with probability 1, and from 1, choose 0 with probability 1.

**Reading reference: which convergence needs which assumptions?** A finite
homogeneous irreducible chain has a unique invariant probability vector
$\boldsymbol\pi$. If it is also aperiodic, then
$\mu_n(j)\to\pi_j$ for every state $j$ and every initial law. Irreducibility
alone gives the time-average result
$n^{-1}\sum_{t=1}^n\mathbf1_{\{X_t=j\}}\to\pi_j$ almost surely;
aperiodicity is not required for this latter statement. These are distinct
finite-chain theorems, not conclusions from the plots. They do not directly
apply to a changing-temperature kernel or assert continuous-state convergence.
The flip example isolates the missing aperiodicity and proves its time-average
limit below without invoking either theorem. No additional exercise is assigned.

Predict the visit fraction when $n$ is even. From start 0,
$v_n=\lceil n/2\rceil/n$; from start 1, $v_n=\lfloor n/2\rfloor/n$.
Thus both visit fractions tend to $1/2$ despite the alternating fixed-time
laws. Here $\lceil z\rceil$ and $\lfloor z\rfloor$ are the least integer
at least $z$ and the greatest integer at most $z$, respectively, for $z\in\mathbb R$.
For either start, $|v_n-1/2|\leq1/(2n)$, which proves this limit directly.

The left graph shows exact laws at times 0 through 16. The right graph
shows only the first 40 visit fractions, to make the alternating pattern
visible; the printed fractions still use 1999 and 2000 transitions.
Each marker is at an integer time and the dashed lines mark $\pi^F_1=1/2$.
With a fixed start this flip path is deterministic; do not infer statistical
dependence merely because its states alternate. With invariant random
initialization, adjacent states are dependent and the process is stationary.

```python
F = np.array([[0., 1.], [1., 0.]])
pi_flip = np.array([0.5, 0.5])
flip_laws = exact_laws(F, [1, 0], law_horizon)
flip_paths = [simulate_two_state(F, start, uniform_inputs[start]) for start in (0, 1)]
flip_fractions = [visit_fractions(path) for path in flip_paths]
print('Invariant law times F:', pi_flip @ F)
print('First six exact laws from fixed start 0:\n', flip_laws[:6])
for start in (0, 1):
    print(f'Flip path, start {start}: v_1999={flip_fractions[start][1998]:.6f}, '
          f'v_2000={flip_fractions[start][1999]:.6f}')
assert np.allclose(pi_flip @ F, pi_flip)
assert np.array_equal(flip_paths[0], np.arange(transition_count+1) % 2)

fig_flip, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].plot(times, flip_laws[:, 1], 'o-')
axes[0].axhline(pi_flip[1], color='black', linestyle='--')
axes[0].set(xlabel='Time t', ylabel='Exact probability Pr(X_t = 1)',
            title='Fixed-time laws alternate', ylim=(-0.02, 1.02))
for start in (0, 1):
    axes[1].plot(sample_counts[:40], flip_fractions[start][:40], 'o-',
                 markersize=3, label=f'Start {start}')
axes[1].axhline(pi_flip[1], color='black', linestyle='--')
axes[1].set(xlabel='Observed transitions n', ylabel='Observed visit fraction v_n',
            title='Time averages tend to one half', ylim=(-0.02, 1.02))
axes[1].legend()
fig_flip.tight_layout()
plt.show()
```

### Optional reading reference: finite cooling and the Markov property

Return temporarily to Session 5's **five-state** set $D=\{0,1,2,3,4\}$;
let $X_t$ now denote that process. Keep the objective $f:D\to\mathbb R$
and proposal law $q$ fixed. Let $(T_t)_{t\geq1}$ be a prescribed deterministic
schedule with $T_t>0$ for every integer $t\geq1$: transition $t$ updates
$X_{t-1}$ to $X_t$ using $K_{T_t}$. Proposal and acceptance inputs are fresh,
mutually independent within and across steps, and independent of $X_0$.

For every integer $t\geq1$, history $(x_0,\ldots,x_{t-1})\in D^t$
with positive probability, and $y\in D$, put $x=x_{t-1}$. Then
$$\begin{aligned}
\Pr(X_t=y\mid X_0=x_0,\ldots,X_{t-1}=x)
 &=\Pr(X_t=y\mid X_{t-1}=x)\\
 &=K_{T_t}(x,y).
\end{aligned}$$
At the specified time $t$, $T_t$ is already known from the schedule. Given
current state $x$, the next update uses only $x$, this temperature, and fresh
inputs; earlier states give no additional information. This is the
**Markov property**, which permits explicit dependence on time.
**Time homogeneity** requires one matrix for all steps.

For the same current state 4 and possible next state 3, the proposal remains
$q(3\mid4)=1/2$ and the objective increase remains $f(3)-f(4)=4-2=2$.
Temperature changes the **complete** transition probability:
$$K_2(4,3)=\tfrac12e^{-1}\approx0.183940,\qquad
  K_{0.5}(4,3)=\tfrac12e^{-4}\approx0.009158.$$
Take $X_0=4$ with probability 1, $T_1=2$, and $T_2=0.5$. Since
$$\Pr(X_1=4\mid X_0=4)=K_2(4,4)=1-\tfrac12e^{-1}>0,$$
both conditional probabilities in the following comparison are defined:
$$\Pr(X_1=3\mid X_0=4)=\tfrac12e^{-1}
\ne\tfrac12e^{-4}=\Pr(X_2=3\mid X_1=4).$$
These condition on the **same current state 4 at different steps**; they
are not a comparison of the marginal probabilities $\Pr(X_1=3)$ and
$\Pr(X_2=3)$. This process is Markov but not time-homogeneous.

Changing temperature **generally** changes $K_T$, not $q$. It need not change
every entry, and need not change the matrix at all: if $f$ is constant, every
feasible tentative proposal has acceptance probability 1 and
$K_T(x,y)=q(y\mid x)$ for every $T>0$. A history-dependent temperature
choice requires a separate analysis; the argument above uses a prescribed schedule.

Writing $\boldsymbol\mu_t$ for the row law of this five-state process gives
$$\boldsymbol\mu_n=\boldsymbol\mu_0 K_{T_1}K_{T_2}\cdots K_{T_n},
\qquad n\in\{1,2,\ldots\}.$$
A single power $P^n$ applies when all transition matrices equal the same $P$,
as in this notebook's separate two-state example. This worked reference
adds no coding experiment.

## E. A continuous transition kernel (5 minutes)

**Reuse S6.5(a); count the explanation once.** Read the supplied two
interval probabilities, compare them at the same current state, and run
this section's code. The numerical helper is supplied: no implementation,
new trajectory, or additional written assignment is required. The third
row, changing scale, is a supplied illustration. The formulas below are
also available as reading references.

Return to Session 5's continuous example
$$D_c=[-2,2],\qquad f_c(x)=(x^2-1)^2+0.2x\quad(x\in D_c).$$
For current $x\in D_c$, scale $s>0$, and temperature $T>0$, propose
$Y=x+sZ$ with $Z\sim\mathcal N(0,1)$. Its **raw density** is
$$q_s(y\mid x)=\frac{1}{s\sqrt{2\pi}}
 e^{-(y-x)^2/(2s^2)},\qquad y\in\mathbb R.$$
For $y\in D_c$, set $\Delta_c(x,y)=f_c(y)-f_c(x)$ and
$\alpha_T^c(x,y)=\min\{1,e^{-\Delta_c(x,y)/T}\}$; accept when an
independent $U\sim\mathrm{Uniform}(0,1)$ is below this threshold.
Reject an infeasible raw candidate before evaluating its objective.

**The complete next-state law.** Let $B\in\mathcal B(D_c)$, where
$\mathcal B(D_c)$ is the sigma-algebra generated by the relatively open
subsets of $D_c$. Write $\mathbf1_B(x)=1$ if $x\in B$ and 0 otherwise.
The transition kernel assigns a next-state probability to each such set:
$$K^c_{T,s}(x,B)=\int_B q_s(y\mid x)\alpha_T^c(x,y)\,dy
                 +r_{T,s}(x)\mathbf1_B(x),$$
$$r_{T,s}(x)=1-\int_{D_c}q_s(y\mid x)\alpha_T^c(x,y)\,dy.$$
Here $dy$ denotes integration with respect to length (Lebesgue measure).
The first term is accepted-move mass; the second is the **rejection atom**
at the current state. The total rejection probability $r_{T,s}(x)$ is
independent of the destination set $B$: its integral covers **all** of $D_c$.
Equivalently,
$$r_{T,s}(x)=\int_{\mathbb R\setminus D_c}q_s(y\mid x)\,dy
 +\int_{D_c}q_s(y\mid x)[1-\alpha_T^c(x,y)]\,dy.$$
The terms cover infeasible proposals and rejected feasible proposals. For
$x\in B$, the complement formula
$K^c_{T,s}(x,B)=1-\int_{D_c\setminus B}q_s\alpha_T^c\,dy$
also holds, but it describes the **whole next-state probability**, not
rejection alone. Thus the raw density alone is not the transition kernel.
For fixed $y=0.8$, $\Pr(Y=y)=0$, even though $q_s(y\mid1)>0$.
The interval $A=[0.75,0.85]$ has positive transition probability. Since
$1\notin A$, its probability from 1 is just the integral, with no atom.
By contrast $K^c_{T,s}(1,\{1\})=r_{T,s}(1)>0$.

**Markov property and time homogeneity.** Let $(X_t)_{t\geq0}$ be the
continuous-state process, with $X_0=1$ almost surely. Prescribe deterministic
$s_t,T_t>0$, and use inputs $(Z_t,U_t)_{t\geq1}$ mutually independent
within and across updates and independent of $X_0$, with the laws above.
Let $\mathcal H_{t-1}=\sigma(X_0,\ldots,X_{t-1})$: the smallest
sigma-algebra making all the recorded states measurable. It represents
information from the entire state history, for example whether a region
was visited. The symbol $\sigma(\cdot)$ means the generated sigma-algebra,
not a sum. For every $t\geq1$ and $B\in\mathcal B(D_c)$,
$$\Pr(X_t\in B\mid\mathcal H_{t-1})
  =K^c_{T_t,s_t}(X_{t-1},B)\quad\text{almost surely}.$$
“Almost surely” means outside an event of probability zero. Conditional
probability here is the conditional expectation of the event indicator;
it does not divide by the probability of an exact continuous history. At a specified time, the current state and prescribed
parameters suffice: the process is Markov. Fixed $s,T$ give a homogeneous
chain; prescribed changes generally give different kernels. History-adaptive
rules require separate analysis; including the relevant adaptation
variables in the state may restore a Markov description.

Keep $s_t=0.2$, with $T_1=0.1$ and $T_2=0.02$. On all of $A$,
$f_c(y)\geq f_c(0.85)=0.24700625>f_c(1)=0.2$. The raw density is positive,
so decreasing $T$ strictly decreases the integrand throughout $A$:
$$K^c_{0.1,0.2}(1,A)>K^c_{0.02,0.2}(1,A).$$
Writing the integral explicitly, for $T>0$,
$$K^c_{T,0.2}(1,A)=\frac{1}{0.2\sqrt{2\pi}}
\int_{0.75}^{0.85}\exp\!\left[-\frac{(y-1)^2}{0.08}
-\frac{(y^2-1)^2+0.2y-0.2}{T}\right]\,dy.$$
The inequality is exact; numerical integration gives approximately
$0.052366$ and $0.003126$. At the first update,
$\Pr(X_1=1\mid X_0=1)=r_{0.1,0.2}(1)\geq\Pr(Z_1>5)>0$
because $1+0.2Z_1>2$ is rejected. As $X_0=1$ almost surely, this also
implies $\Pr(X_1=1)>0$.
Consequently both $\Pr(X_1\in A\mid X_0=1)$ and
$\Pr(X_2\in A\mid X_1=1)$ are defined and equal the respective kernel
values above. These are unequal probabilities conditioned on the **same
current state 1 at different times**, not unequal marginal laws.
The process is Markov and is not homogeneous.

**Read the supplied computation.** Gauss–Legendre quadrature approximates
an integral by a deterministic weighted sum; it is not Monte Carlo sampling.
The helper is restricted to start 1 and subintervals of $A$, where the
integrand is smooth and the current-state atom is absent. It does not
integrate the full domain or compute $r_{T,s}(1)$. The plotted vertical
quantity is $q_s(y\mid1)\alpha_T^c(1,y)$, an accepted-move **density**.
Area over $A$ is the interval probability; a curve height is not the
probability of a single point. The third supplied row changes only $s$
from 0.2 to 0.1 at $T=0.1$, showing that scale also changes the complete kernel.

```python
def f_c(values):
    values = np.asarray(values, dtype=float)
    if np.any(~np.isfinite(values)) or np.any((values < -2) | (values > 2)):
        raise ValueError('f_c is defined only on [-2, 2].')
    return (values**2-1)**2 + 0.2*values

def raw_gaussian_density(y, x, scale):
    y = np.asarray(y, dtype=float)
    if (not np.isfinite(x) or not -2 <= x <= 2 or not np.isfinite(scale)
            or scale <= 0 or np.any(~np.isfinite(y))):
        raise ValueError('Use feasible x, finite positive scale, and finite raw locations.')
    return np.exp(-0.5*((y-x)/scale)**2)/(scale*np.sqrt(2*np.pi))

def accepted_move_density(y, x, scale, temperature):
    if not np.isfinite(temperature) or temperature <= 0:
        raise ValueError('Use a finite positive temperature.')
    # f_c validates feasibility; this expression describes only accepted moves.
    delta = f_c(y)-f_c(x)
    return raw_gaussian_density(y, x, scale)*np.exp(-np.maximum(delta, 0)/temperature)

def accepted_interval_mass(x, scale, temperature, left=0.75, right=0.85, order=64):
    # Deliberately restricted: A excludes current x and has a smooth positive-delta integrand.
    if (not all(np.isfinite(v) for v in (x, scale, temperature, left, right))
            or x != 1 or scale <= 0 or temperature <= 0
            or not 0.75 <= left <= right <= 0.85):
        raise ValueError('Use x=1, positive s/T, and a subinterval of [0.75, 0.85].')
    if not isinstance(order, (int, np.integer)) or order < 2:
        raise ValueError('Quadrature order must be an integer of at least 2.')
    if left == right:
        return 0.0  # A singleton distinct from current x has zero transition mass.
    nodes_c, weights_c = np.polynomial.legendre.leggauss(order)
    locations_c = (left+right)/2 + (right-left)*nodes_c/2
    return float((right-left)/2*np.dot(weights_c,
                 accepted_move_density(locations_c, x, scale, temperature)))

# Supplied deterministic quadratures: no random input or new path is generated.
continuous_kernel_settings = [('First temperature', 0.2, 0.1),
                              ('Second temperature', 0.2, 0.02),
                              ('Change only scale', 0.1, 0.1)]
continuous_kernel_values = [accepted_interval_mass(1., scale_c, temperature_c)
                            for _, scale_c, temperature_c in continuous_kernel_settings]
print('Numerical interval probabilities from x=1 into A=[0.75, 0.85]')
print('Gauss-Legendre order 64; numerical approximations, not sampled frequencies')
print(f"{'Case':<21} {'s':>6} {'T':>6} {'K(1,A)':>16}")
for (case_name, scale_c, temperature_c), interval_mass_c in zip(
        continuous_kernel_settings, continuous_kernel_values):
    print(f'{case_name:<21} {scale_c:6.2f} {temperature_c:6.2f} {interval_mass_c:16.12f}')
print('Singleton {0.8} from current state 1:',
      accepted_interval_mass(1., 0.2, 0.1, left=0.8, right=0.8))
print('The atom at current state 1 is excluded from A and is not computed here.')

fig_continuous_kernel, kernel_axis = plt.subplots(figsize=(7.2, 3.8))
interval_grid_c = np.linspace(0.75, 0.85, 301)
for index_c, (_, scale_c, temperature_c) in enumerate(continuous_kernel_settings):
    density_c = accepted_move_density(interval_grid_c, 1., scale_c, temperature_c)
    curve_c, = kernel_axis.plot(interval_grid_c, density_c,
                               linestyle='--' if index_c == 2 else '-',
                               label=f's={scale_c}, T={temperature_c}')
    if index_c < 2:
        kernel_axis.fill_between(interval_grid_c, density_c, color=curve_c.get_color(), alpha=0.10)
kernel_axis.set(xlabel='Feasible next location y', ylabel='Accepted-move density',
                title='Accepted moves into A = [0.75, 0.85]', xlim=(0.75, 0.85), ylim=(0, None))
kernel_axis.legend()
fig_continuous_kernel.tight_layout()
plt.show()
```
